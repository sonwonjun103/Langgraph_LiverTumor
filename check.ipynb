```python
from totalsegmentator.python_api import totalsegmentator

def extract_segmentation(input_path, output_path):
    totalsegmentator(
        input=input_path,
        output=output_path,
        ml=False,
        fast=True,
        verbose=False,
        roi_subset=["liver"],
    )


extract_segmentation(
    input_path = f"/mnt/d/research/Liver/register_YU/11/register/A.nii.gz",
    output_path = f"/mnt/d/research/Liver/register_YU/11/liver",
)
```
